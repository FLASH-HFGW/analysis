```python
import numpy as np
import matplotlib.pyplot as plt

# Nome del file .npz
file_npz = "run00359_event00000010_fft_summary.npz"   # cambia con il tuo file

# Quale evento/spettro vuoi plottare
idx = 0   # primo evento/spettro

with np.load(file_npz) as data:

    fft_amp_mode0_SPECs = data["fft_amp_mode0_SPECs"]
    fft_freq_mode0 = np.linspace(data["fft_freq_mode0"][0], data["fft_freq_mode0"][1], int(data["fft_freq_mode0"][2]-1))
    print(len(fft_amp_mode0_SPECs), len(fft_freq_mode0))

    fft_amp_mode1_SPECs = data["fft_amp_mode1_SPECs"]
    fft_freq_mode1 = np.linspace(data["fft_freq_mode1"][0], data["fft_freq_mode1"][1], int(data["fft_freq_mode1"][2]-1))

    fft_amp_mode2_SPECs = data["fft_amp_mode2_SPECs"]
    fft_freq_mode2 = np.linspace(data["fft_freq_mode2"][0], data["fft_freq_mode2"][1], int(data["fft_freq_mode2"][2]-1))

    print("Shape mode 0 amp:", fft_amp_mode0_SPECs.shape)
    print("Shape mode 0 freq:", fft_freq_mode0.shape)

    print("Shape mode 1 amp:", fft_amp_mode1_SPECs.shape)
    print("Shape mode 1 freq:", fft_freq_mode1.shape)

    print("Shape mode 2 amp:", fft_amp_mode2_SPECs.shape)
    print("Shape mode 2 freq:", fft_freq_mode2.shape)


# Funzione comoda per prendere lo spettro giusto
def get_spectrum(amp, freq, idx=0):
    """
    Se amp è 2D, prende amp[idx].
    Se amp è 1D, usa direttamente amp.
    Stessa cosa per freq.
    """

    amp = np.asarray(amp)
    freq = np.asarray(freq)

    if amp.ndim == 1:
        y = amp
    else:
        y = amp[idx]

    if freq.ndim == 1:
        x = freq
    else:
        x = freq[idx]

    return x, y


# Prendo i tre spettri FFT
f0, A0 = get_spectrum(fft_amp_mode0_SPECs, fft_freq_mode0, idx)
f1, A1 = get_spectrum(fft_amp_mode1_SPECs, fft_freq_mode1, idx)
f2, A2 = get_spectrum(fft_amp_mode2_SPECs, fft_freq_mode2, idx)


# Plot dei 3 modi
plt.figure(figsize=(12, 8))

plt.subplot(3, 1, 1)
plt.plot(f0, A0)
plt.xlabel("Frequenza [Hz]")
plt.ylabel("Ampiezza [mW]")
plt.title("FFT - mode 0")
plt.yscale("log")
plt.grid(True)

plt.subplot(3, 1, 2)
plt.plot(f1, A1)
plt.xlabel("Frequenza [Hz]")
plt.ylabel("Ampiezza [mW]")
plt.title("FFT - mode 1")
plt.yscale("log")
plt.grid(True)

plt.subplot(3, 1, 3)
plt.plot(f2, A2)
plt.xlabel("Frequenza [Hz]")
plt.ylabel("Ampiezza [mW]")
plt.title("FFT - mode 2")
plt.yscale("log")
plt.grid(True)


plt.tight_layout()
plt.show()
```
